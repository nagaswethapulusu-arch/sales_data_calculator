# Sales Data Calculator
This notebook calculates the total, average, highest and lowest sales.

```python
days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7", "Day 8", "Day 9", "Day 10"]
sales = [12500, 8900, 15200, 9800, 17600, 11300, 14100, 7600, 16400, 13200]
```

```python
total = sum(sales)
number_of_sales = len(sales)
average = total / number_of_sales
highest = max(sales)
lowest = min(sales)
best_day = days[sales.index(highest)]
worst_day = days[sales.index(lowest)]
```

```python
print(f"Total sales: ₹{total:,}")
print(f"Number of sales days: {number_of_sales}")
print(f"Average sales: ₹{average:,.2f}")
print(f"Highest sale: ₹{highest:,} ({best_day})")
print(f"Lowest sale: ₹{lowest:,} ({worst_day})")
```

```python
above_average = [s for s in sales if s > average]
print(f"Days above average: {len(above_average)} out of {number_of_sales}")
```
